```python
# Practice 1: Student Information System
'''Create a class named Student'''
class Student:

    '''Constructor'''
    def __init__(self, name, age, student_id):
        self.name = name
        self.age = age
        self.student_id = student_id

    '''Create an instance method named details()'''
    def details(self):
        print("Student Name:", self.name)
        print("Age:", self.age)
        print("Student ID:", self.student_id)

    '''Create an instance method named is_adult'''
    def is_adult(self):
        if self.age < 0:
            print("Invalid input")
        elif self.age >= 18:
            adult = "Yes"
        else:
            adult = "No"
        print("Adult:", adult)

'''Create student 1'''
print("\nStudent 1")
student1 = Student("Makara", 20, "e20240242")
student1.details()
student1.is_adult()

'''Create student 2'''
print("\nStudent 2")
student2 = Student("Pisey", 15, "e20240986")
student2.details()
student2.is_adult()

'''Create student 3'''
print("\nStudent 3")
student3 = Student("Piseth", 23, "e20240472")
student3.details()
student3.is_adult()
```
