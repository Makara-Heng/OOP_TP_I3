```python
# Create a class Book with title, author, and price attributes.
# Demonstrate encapsulation by protecting the price attribute.
# Use inheritance to create an EBook subclass.
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price     #private(need getter and setter)

    def get_price(self):         #getter
        return self.__price
    
    def set_price(self, price):  #setter
        self.__price = price

class EBook(Book):
    def __init__(self, title, author, price):
        super().__init__(title, author, price)

# Get user input
print("Enter Book Information")
title = input("Enter book title: ")
author = input("Enter author name: ")
price = float(input("Enter book price: "))

# Create Book object
book1 = Book(title, author, price)

# Display Book information
print("\n---Book Information---")
print("Title:", book1.title)
print("Author:", book1.author)
print("Price:", book1.get_price())

# Get EBook input
print("\nEnter EBook Information")
ebook_title = input("Enter ebook title: ")
ebook_author = input("Enter author name: ")
ebook_price = float(input("Enter ebook price: "))

# Create EBook object
ebook1 = EBook(ebook_title, ebook_author, ebook_price)

# Display EBook information
print("\n---EBook Information---")
print("Title:", ebook1.title)
print("Author:", ebook1.author)
print("Price:", ebook1.get_price())
```
