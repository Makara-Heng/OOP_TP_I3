```python
# Practice 4: Employee Management
'''Create class named Employee'''
class Employee:
    company = "ITC"
    employee_count = 0

    '''Constructor'''
    def __init__(self, name, position, salary):
        self.name = name
        self.position = position
        self.salary = salary

        '''Increase the employee count whenever an object is created'''
        Employee.employee_count += 1

    '''Create an instance method display_info()'''
    def display_info(self):
        print("\nEmployee:", self.name)
        print("Position:", self.position)
        print("Salary:", self.salary)

    '''Create a class method'''
    @classmethod
    def get_employee_count(cls):
        return cls.employee_count

'''Create at least four employees'''
emp1 = Employee("Kimheng", "Manager", 1000)
emp2 = Employee("Ching", "Engineer", 1200)
emp3 = Employee("Sivchhoeng", "Assistant", 1100)
emp4 = Employee("Makara", "HR", 1500)

'''Display all employee information and total number of employees'''
print(f"Company: {Employee.company}")

emp1.display_info()
emp2.display_info()
emp3.display_info()
emp4.display_info()

print(f"\nTotal Employees: {Employee.get_employee_count()}")
```
