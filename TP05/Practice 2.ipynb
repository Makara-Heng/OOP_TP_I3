```python
# Practice 2: Bank Account
'''Create a class named BankAccount'''
class BankAccount:
    bank_name = "ITC Bank"

    '''Constructor'''
    def __init__(self, owner, account_number, balance):
        self.owner = owner
        self.account_number = account_number
        self.balance = balance

    '''Create method named deposit(amount)'''
    def deposit(self, amount):
        if amount > 0:
            print(f"\nDeposit: ${amount}")
            self.balance += amount
        else:
            print("Invalid number!")

    '''Create method named withdraw(amount)'''
    def withdraw(self, amount):
        if amount < 0:
            print("Invalid number!")
        elif self.balance < amount:
            print("Withdrawal must be less than account balance.")
        else:
            print(f"\nWithdraw: ${amount}")
            self.balance -= amount

    '''Create method name display_balance()'''
    def display_balance(self):
        print("Bank:", BankAccount.bank_name)
        print("Account Owner:", self.owner)
        print(f"Initial Balance: ${self.balance}")

'''Create Account 1'''
account1 = BankAccount("Makara", 1, 2000)

print("Account 1")
account1.display_balance()

account1.deposit(500)
print(f"New Balance: ${account1.balance}")

account1.withdraw(200)
print(f"Final Balance: ${account1.balance}")

print("===========================")

'''Create Account 2'''
account2 = BankAccount("Pisey", 2, 2500)

print("Account 2")
account2.display_balance()

account2.deposit(300)
print(f"New Balance: ${account2.balance}")

account2.withdraw(200)
print(f"Final Balance: ${account2.balance}")
```
