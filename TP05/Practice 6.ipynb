```python
# Practice 6: Product Management System
'''Create a class named Product'''
class Product:
    store_name = "ITC Online Store"
    product_count = 0

    '''Constructor'''
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

        '''Increase the product count when a new object is created'''
        Product.product_count += 1

    '''Create an instance method display_product()'''
    def display_product(self):
        print(f"Name: {self.name}")
        print(f"Price: ${self.price}")
        print(f"Quantity: {self.quantity}")
        print(f"Total Value: ${self.total_value()}\n")

    '''Create an instance method total_value()'''
    def total_value(self):
        return self.price * self.quantity
    
    '''Create a class method get_product_count()'''
    @classmethod
    def get_product_count(cls):
         return cls.product_count

    '''Create a static method is_valid_price(price)'''
    @staticmethod
    def is_valid_price(price):
        return price > 0

'''Product 1'''
if Product.is_valid_price(800):
    product1 = Product("Laptop", 800, 5)
    print("==========ITC ONLINE STORE==========\n")
    print("Product 1")
    product1.display_product()

'''Product 2'''
if Product.is_valid_price(15):
    product2 = Product("Mouse", 15, 20)
    print("Product 2")
    product2.display_product()

'''Product 3'''
if Product.is_valid_price(30):
    product3 = Product("Keyboard", 30, 10)
    print("Product 3")
    product3.display_product()

print(f"Total Products: {Product.get_product_count()}")

total_inventory = (product1.total_value() + product2.total_value() + product3.total_value())
print(f"Total Inventory Value: ${total_inventory}") 
```
