```python
# Practice 5: Temperature Converter
'''Create class named Temperature'''
class Temperature:

    '''Create a static method celsius_to_fahrenheit(celsius)'''
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (9/5)*celsius + 32

    '''Create a static method fahrenheit_to_celsius(fahrenheit)'''
    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        return (5/9)*(fahrenheit-32)

    '''Create a static method celsius_to_kelvin(celsius)'''
    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15

'''Test and display all the conversion results'''
c = 25
print("Celsius:", c)
print("Fahrenheit:", Temperature.celsius_to_fahrenheit(c))
print("Kelvin:", Temperature.celsius_to_kelvin(c))

c = 100
print("\nCelsius:", c)
print("Fahrenheit:", Temperature.celsius_to_fahrenheit(c))
print("Kelvin:", Temperature.celsius_to_kelvin(c))

c = 37
print("\nCelsius:", c)
print("Fahrenheit:", Temperature.celsius_to_fahrenheit(c))
print("Kelvin:", Temperature.celsius_to_kelvin(c))
```
